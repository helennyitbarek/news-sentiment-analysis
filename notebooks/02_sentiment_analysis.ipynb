# 📊 News Sentiment Analysis – Exploratory Data Analysis (EDA)

## 📁 Dataset Overview
This dataset contains financial news articles with the following key columns:
- **headline**: News headline text
- **url**: Source link
- **publisher**: News author or outlet
- **date**: Publication timestamp
- **stock**: Associated stock ticker

---

## 🧹 Data Cleaning Steps
- Loaded dataset from CSV file
- Converted `date` column to datetime using `pd.to_datetime()` with `errors='coerce'`
- Removed rows with invalid or missing dates using `dropna()`
- Ensured clean time-series data for analysis

---

## 📅 Time Feature Processing
The `date` column was processed to enable time-based analysis such as trends over time, frequency of news, and publication patterns.

---

## 📊 Key Observations
- News volume varies significantly over time
- A small number of publishers contribute most of the content
- Some stocks appear more frequently in news headlines than others
- Headlines vary in length and structure

---

## ✅ Conclusion
This EDA provides an initial understanding of financial news patterns and prepares the dataset for deeper sentiment analysis and modeling.
